```python
df["TotalCharges"]= pd.to_numeric(df['TotalCharges'], errors = 'coerce')
```

```python
print("Missing after convert:",df['TotalCharges'].isnull().sum())
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
```

```python
print("Missing :",df['TotalCharges'].isnull().sum())
```

```python
print("dtype now:",df['TotalCharges'].dtypes)
```

```python
df = df.drop(columns = ['customerID'])
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1,'No':0})
df['gender'] = df['gender'].map({'Male': 1,'Female':0})
```

```python
cat_cols = df.select_dtypes(include = ['object']).columns.tolist()
print(cat_cols)
```

```python
for col in cat_cols:
    print(f"{col}:{df[col].unique()}")
```

```python
for col in cat_cols:
    if df[col].nunique() == 2:
        print(f"{col} ->Label Encoding")
    else:
        print(f"{col} ->One Hot Encoding")
```

```python
df = pd.get_dummies(df, columns=[
    'InternetService', 'Contract',
    'PaymentMethod', 'MultipleLines',
    'OnlineSecurity', 'OnlineBackup',
    'DeviceProtection', 'TechSupport',
    'StreamingTV', 'StreamingMovies'
])
print(df.shape)
print(df.dtypes)
```

Fix- Remaining str columns encode

```python
binary_cols = [col for col in df.columns if df[col].dtypes == 'str' and df[col].nunique() == 2]
print ("Binary columns: ",binary_cols)
```

```python
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})
print(df.dtypes)
print(df.isnull().sum())
```
